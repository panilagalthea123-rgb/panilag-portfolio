# Home 1.1 — Text Mining Activity: Deep Learning

This activity explores the foundation of deep learning and recent advances in the field.


## 1. What is the foundation/concept of the idea?

The foundation of deep learning is artificial neural networks and representation learning.

The basic idea is to imitate, in a simplified mathematical way, how interconnected neurons process information. A neural network consists of layers of connected artificial neurons. During training, the network adjusts its weights using algorithms such as backpropagation and gradient-based optimization so that it can learn patterns from data.

The important concept that makes it deep learning is hierarchical feature/representation learning.

Instead of humans manually defining every feature, multiple layers allow the model to progressively learn more complex representations from the input.

### Hierarchical Representation Learning

The general process can be represented as:

**Raw data → Simple features → Complex features → Representation → Prediction**

For an image:

**Pixels → Edges → Shapes → Objects → Classification**

For text:

**Words/tokens → Relationships → Meaning/context → Language representation → Prediction/generation**

The developments in deep learning are connected because AlexNet, VGG, GoogleNet, ResNet, DenseNet, and GANs are different architectures or approaches built upon the broader foundation of neural networks and deep representation learning.

## 2. What recent/advanced progress in deep learning?

The timeline ends around 2017, so it does not include several important advances that shaped modern deep learning.

These include:

- Transformers
- Large Language Models (LLMs)
- Foundation models
- Diffusion models
- Generative AI
- Multimodal AI
- Retrieval-Augmented Generation (RAG)
- Scientific AI
- Agentic AI

These developments have expanded deep learning from mainly recognizing and classifying patterns to understanding, generating, retrieving, and working with different types of information.

## Recent Developments in Deep Learning

| Year/Period | Development | Importance |
|---|---|---|
| 2017 | Transformers | Introduced self-attention and changed sequence processing |
| 2018 | BERT / modern pretrained language models | Major advancement in language and understanding |
| 2019–2020 | Large-scale language models | Increasing model scale and general capabilities |
| 2020 | Diffusion Models | Major new approach to generative AI |
| 2020 onward | Foundation models | One pretrained model that can support many tasks |
| 2021 onward | Vision Transformers / Multimodal Models | Transformers expand beyond text |
| 2022 onward | Generative AI boom | Large-scale public use of text/image generation |
| 2023 onward | RAG | Combine LLMs with external knowledge |
| 2023 onward | Multimodal AI | Text, image, audio and video integration |
| 2024 onward | Scientific AI Models | Deep learning increasingly applied to scientific discovery |
| 2025–2026 | Agentic & Multimodal AI | Models increasingly retrieve information, use tools, and operate across modalities |
