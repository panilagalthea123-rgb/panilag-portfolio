**Laboratory Task 2**

Instruction: Perform a single forward pass and compute for the error.

Given:

x = [1, 0, 1]
y = 1

Hidden layer weights:

w11 = 0.2
w12 = -0.3
w13 = 0.4
w14 = 0.1
w15 = -0.5
w16 = 0.2

Output layer weights:

w21 = -0.3
w22 = -0.2

Biases:

θ1 = -0.4
θ2 = 0.2
θ3 = 0.1


1. Calculate Z1

Z1 = (w11 × x1) + (w13 × x2) + (w15 × x3) + θ1

Z1 = (0.2 × 1) + (0.4 × 0) + (-0.5 × 1) + (-0.4)

Z1 = 0.2 + 0 - 0.5 - 0.4

Z1 = -0.7


2. Calculate Z2

Z2 = (w12 × x1) + (w14 × x2) + (w16 × x3) + θ2

Z2 = (-0.3 × 1) + (0.1 × 0) + (0.2 × 1) + 0.2

Z2 = -0.3 + 0 + 0.2 + 0.2

Z2 = 0.1


3. Apply the sigmoid activation function

Sigmoid function:

σ(z) = 1 / (1 + e^(-z))

H1 = σ(Z1)

H1 = σ(-0.7)

H1 ≈ 0.3318


H2 = σ(Z2)

H2 = σ(0.1)

H2 ≈ 0.5250


4. Calculate the output weighted sum

Z3 = (w21 × H1) + (w22 × H2) + θ3

Z3 = (-0.3 × 0.3318) + (-0.2 × 0.5250) + 0.1

Z3 = -0.09954 - 0.105 + 0.1

Z3 ≈ -0.10454


5. Apply the ReLU activation function

f = max(0, Z3)

ŷ = max(0, -0.10454)

ŷ = 0


6. Calculate the error

Actual output:

y = 1

Predicted output:

ŷ = 0

Using the squared error formula:

E = 1/2(y - ŷ)^2

E = 1/2(1 - 0)^2

E = 1/2

E = 0.5


Final Answers:

Z1 = -0.7
Z2 = 0.1
H1 ≈ 0.3318
H2 ≈ 0.5250
Z3 ≈ -0.10454
Predicted output (ŷ) = 0
Error = 0.5
