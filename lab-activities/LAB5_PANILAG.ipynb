```python
import torch
import numpy as np
```

```python
def set_seed(seed: int):
    np.random.seed(seed)
    torch.manual_seed(seed)
```

```python
set_seed(42)

arr = np.random.randint(0, 5, 6)

print(arr)
```

```python
x = torch.from_numpy(arr)

print(x)
```

```python
x = x.to(torch.int64)

print(x)
print(x.dtype)
```

```python
x = x.reshape(3, 2)

print(x)
```

```python
right_column = x[:, 1]

print(right_column)
```

```python
squared_x = x ** 2

print(squared_x)
```

```python
set_seed(42)

y = torch.randint(0, 5, (2, 3), dtype=torch.int64)

print(y)
```

```python
product = x @ y

print(product)
```
